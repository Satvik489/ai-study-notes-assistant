```python

from transformers import pipeline

print("Loading AI model...")

ai_model = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct"
)

print("AI model loaded successfully!")
```

```python
def track_progress(student_name, topic, total_questions, correct_answers, history):

    total_questions = int(total_questions)
    correct_answers = int(correct_answers)

    history = list(history or [])

    if total_questions <= 0:
        return history, "Total questions must be greater than zero."

    if correct_answers < 0 or correct_answers > total_questions:
        return history, "Correct answers must be between 0 and the total questions."

    percentage = (correct_answers / total_questions) * 100

    history.append({
        "student": student_name,
        "topic": topic,
        "score": correct_answers,
        "total": total_questions,
        "percentage": percentage
    })

    result = "LEARNING PROGRESS HISTORY\n\n"

    for i, attempt in enumerate(history, 1):
        result += (
            f"Attempt {i}: {attempt['topic']}\n"
            f"Score: {attempt['score']}/{attempt['total']}\n"
            f"Percentage: {attempt['percentage']:.1f}%\n\n"
        )

    return history, result
```

```python

import gradio as gr


def study_assistant(student_name, subject, topic, notes, difficulty, question_count):

    summary = ai_model(
        "Summarise these study notes in simple language: " + notes,
        max_new_tokens=150,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    important_points = ai_model(
        f"List the 5 most important points about {topic} in {subject}.",
        max_new_tokens=150,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    simple_explanation = ai_model(
        f"Explain {topic} in {subject} in very simple language with an example.",
        max_new_tokens=200,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    questions = ai_model(
        f"Create exactly {int(question_count)} numbered practice questions "
        f"about {topic} in {subject} for a Class 12 student. "
        f"Difficulty: {difficulty}. "
        "Write short, clear and complete questions. "
        "Do not provide answers.",
        max_new_tokens=500,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    report = f"""
AI STUDY NOTES ASSISTANT

Student: {student_name}
Subject: {subject}
Topic: {topic}
Difficulty: {difficulty}
Number of Questions: {question_count}

SUMMARY:
{summary}

IMPORTANT POINTS:
{important_points}

SIMPLE EXPLANATION:
{simple_explanation}

PRACTICE QUESTIONS:
{questions}
"""

    report_path = "/content/AI_Study_Report.txt"

    with open(report_path, "w", encoding="utf-8") as file:
        file.write(report)

    return summary, important_points, simple_explanation, questions, report_path



with gr.Blocks() as demo:

    gr.Markdown("# AI Study Notes Assistant – Enhanced Version")
    gr.Markdown("Your personal AI-powered study companion.")

    with gr.Row():
        student_name = gr.Textbox(label="Student Name")
        subject = gr.Dropdown(
            ["Physics", "Chemistry", "Mathematics"],
            label="Select Subject"
        )

    topic = gr.Textbox(label="Enter Topic")
    notes = gr.Textbox(
        label="Paste Your Study Notes",
        lines=6
    )

    difficulty = gr.Dropdown(
        ["Easy", "Medium", "Hard"],
        label="Choose Difficulty",
        value="Medium"
    )

    question_count = gr.Slider(
        minimum=3,
        maximum=10,
        step=1,
        value=5,
        label="Number of Practice Questions"
    )

    generate_button = gr.Button("Generate Study Material")

    summary = gr.Textbox(label="AI Summary")
    important_points = gr.Textbox(label="Important Points")
    explanation = gr.Textbox(label="Simple Explanation")
    questions = gr.Textbox(label="Practice Questions", lines=10)
    report = gr.File(label="Download Your Study Report")

    generate_button.click(
        fn=study_assistant,
        inputs=[
            student_name,
            subject,
            topic,
            notes,
            difficulty,
            question_count
        ],
        outputs=[
            summary,
            important_points,
            explanation,
            questions,
            report
        ]
    )

    gr.Markdown("## 📊 Learning Progress Tracker")

    progress_name = gr.Textbox(label="Student Name")
    progress_topic = gr.Textbox(label="Topic")

    total_questions = gr.Number(
        label="Total Questions",
        value=5,
        precision=0
    )

    correct_answers = gr.Number(
        label="Correct Answers",
        value=0,
        precision=0
    )

    track_button = gr.Button("Record My Progress")
    progress_history = gr.State([])
    progress_result = gr.Textbox(
        label="My Learning Progress",
        lines=5
    )

    track_button.click(
        fn=track_progress,
        inputs=[
            progress_name,
            progress_topic,
            total_questions,
            correct_answers,
            progress_history
        ],
        outputs=[
            progress_history,
            progress_result
        ]
    )

demo.launch(share=True, debug=True)
```

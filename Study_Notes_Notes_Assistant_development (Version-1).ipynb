```python
import ast

search_topic = input("\nEnter a topic to search: ")

found = False

with open("study_notes.txt", "r") as file:
    for line in file:
        saved_note = ast.literal_eval(line)

        if search_topic.lower() in saved_note["topic"].lower():
            print("\n===== NOTE FOUND =====")
            print("Subject:", saved_note["subject"])
            print("Topic:", saved_note["topic"])
            print("Notes:", saved_note["notes"])
            found = True

if not found:
    print("No notes found for this topic.")
```

```python
import re
from collections import Counter

stop_words = {
    "the", "is", "a", "an", "of", "to", "in", "and",
    "with", "for", "on", "are", "it", "deals"
}

words = re.findall(r'\b[a-zA-Z]+\b', notes.lower())

keywords = [
    word for word in words
    if word not in stop_words and len(word) > 2
]

important_words = Counter(keywords).most_common(5)

print("\n===== IMPORTANT KEYWORDS =====")

for word, count in important_words:
    print(word, "-", count)
```

```python

from transformers import pipeline
import re
from collections import Counter
from google.colab import files

# 1. Load AI model
print("Loading AI model...")

ai_model = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct"
)

# 2. Collect student information
student_name = input("Enter your name: ")
subject = input("Enter your subject: ")
topic = input("Enter your topic: ")
notes = input("Enter your study notes: ")

# 3. Organise information
subject = subject.strip().lower()

if subject == "chemistry":
    selected_subject = "Chemistry"
elif subject == "physics":
    selected_subject = "Physics"
elif subject == "maths":
    selected_subject = "Mathematics"
else:
    selected_subject = "Other"

student_notes = {
    "student": student_name,
    "subject": selected_subject,
    "topic": topic,
    "notes": notes
}

# 4. Save notes
with open("study_notes.txt", "a") as file:
    file.write(str(student_notes) + "\n")

# 5. Extract keywords
stop_words = {
    "the", "is", "a", "an", "of", "to", "in",
    "and", "with", "for", "on", "are", "it"
}

words = re.findall(r'\b[a-zA-Z]+\b', notes.lower())

keywords = [
    word for word in words
    if word not in stop_words and len(word) > 2
]

important_words = Counter(keywords).most_common(5)

# 6. Generate AI summary
print("\nGenerating AI summary...")

ai_summary = ai_model(
    "Summarise these notes in simple language: " + notes,
    max_new_tokens=100,
    do_sample=False,
    return_full_text=False
)[0]["generated_text"]

# 7. Generate AI explanation
print("Generating concept explanation...")

explanation = ai_model(
    f"Explain {topic} in {selected_subject} "
    "in simple language for a Class 12 student.",
    max_new_tokens=150,
    do_sample=False,
    return_full_text=False
)[0]["generated_text"]

# 8. Generate practice questions
print("Generating practice questions...")

questions = ai_model(
    f"Generate 5 practice questions for Class 12 "
    f"on {topic} in {selected_subject}.",
    max_new_tokens=200,
    do_sample=False,
    return_full_text=False
)[0]["generated_text"]

# 9. Display results
print("\n===== AI STUDY NOTES ASSISTANT =====")
print("Student:", student_name)
print("Subject:", selected_subject)
print("Topic:", topic)

print("\n===== IMPORTANT KEYWORDS =====")
for word, count in important_words:
    print(word, "-", count)

print("\n===== AI SUMMARY =====")
print(ai_summary)

print("\n===== AI EXPLANATION =====")
print(explanation)

print("\n===== AI PRACTICE QUESTIONS =====")
print(questions)

# 10. Create final report
with open("AI_Study_Report.txt", "w") as file:
    file.write("AI STUDY NOTES ASSISTANT\n\n")
    file.write(f"Student: {student_name}\n")
    file.write(f"Subject: {selected_subject}\n")
    file.write(f"Topic: {topic}\n\n")
    file.write(f"Original Notes:\n{notes}\n\n")
    file.write(f"AI Summary:\n{ai_summary}\n\n")
    file.write(f"AI Explanation:\n{explanation}\n\n")
    file.write(f"Practice Questions:\n{questions}\n\n")
    file.write("Important Keywords:\n")

    for word, count in important_words:
        file.write(f"{word}: {count}\n")

print("\nComplete AI Study Report generated!")

# 11. Download report
files.download("AI_Study_Report.txt")
```

```python

import gradio as gr

def study_assistant(student_name, subject, topic, notes):

    summary = ai_model(
        "Summarise these study notes in simple language: " + notes,
        max_new_tokens=150,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    important_points = ai_model(
        f"List the 5 most important points about {topic} in {subject}.",
        max_new_tokens=150,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    simple_explanation = ai_model(
        f"Explain {topic} in {subject} in very simple language with an example.",
        max_new_tokens=200,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    questions = ai_model(
        f"Generate 5 practice questions on {topic} in {subject}.",
        max_new_tokens=200,
        do_sample=False,
        return_full_text=False
    )[0]["generated_text"]

    report = f"""
AI STUDY NOTES ASSISTANT

Student: {student_name}
Subject: {subject}
Topic: {topic}

SUMMARY:
{summary}

IMPORTANT POINTS:
{important_points}

SIMPLE EXPLANATION:
{simple_explanation}

PRACTICE QUESTIONS:
{questions}
"""

    report_path = "/content/AI_Study_Report.txt"

    with open(report_path, "w", encoding="utf-8") as file:
        file.write(report)

    return summary, important_points, simple_explanation, questions, report_path


demo = gr.Interface(
    fn=study_assistant,
    inputs=[
        gr.Textbox(label="Student Name"),
        gr.Dropdown(
            ["Physics", "Chemistry", "Mathematics"],
            label="Select Subject"
        ),
        gr.Textbox(label="Enter Topic"),
        gr.Textbox(label="Paste Your Study Notes", lines=6)
    ],
    outputs=[
        gr.Textbox(label="AI Summary"),
        gr.Textbox(label="Important Points"),
        gr.Textbox(label="Simple Explanation"),
        gr.Textbox(label="Practice Questions")
    ],
    title="AI Study Notes Assistant",
    description="Your personal AI-powered study companion."
)

demo.launch(share=True)

demo = gr.Interface(
    fn=study_assistant,
    inputs=[
        gr.Textbox(label="Student Name"),
        gr.Dropdown(
            ["Physics", "Chemistry", "Mathematics"],
            label="Select Subject"
        ),
        gr.Textbox(label="Enter Topic"),
        gr.Textbox(label="Paste Your Study Notes", lines=6)
    ],
    outputs=[
    gr.Textbox(label="AI Summary"),
    gr.Textbox(label="Important Points"),
    gr.Textbox(label="Simple Explanation"),
    gr.Textbox(label="Practice Questions"),
    gr.File(label="Download Your Study Report")
],
    title="AI Study Notes Assistant",
    description="Your personal AI-powered study companion."
)

demo.launch(share=True)
```

# AI Study Notes Assistant
A simple assistant for recording and saving student study notes.

```python
student_name = input("Enter your name: ").strip()
subject_input = input("Enter your subject: ").strip().lower()

subject_map = {
    "chemistry": "Chemistry",
    "physics": "Physics",
    "maths": "Mathematics",
    "mathematics": "Mathematics",
}

selected_subject = subject_map.get(subject_input, "Other")

print("Welcome", student_name)
print("Selected subject:", selected_subject)
```

```python
topic = input("Enter the topic name: ").strip()
notes = input("Enter your study notes: ").strip()

print("\n===== AI STUDY NOTES ASSISTANT =====")
print("Student:", student_name)
print("Subject:", selected_subject)
print("Topic:", topic)
print("Your Notes:", notes)
```

```python
word_count = len(notes.split())
print("Total words in your notes:", word_count)
```

```python
student_notes = {
    "student": student_name,
    "subject": selected_subject,
    "topic": topic,
    "notes": notes,
}

print("\n===== SAVED NOTE =====")
for key, value in student_notes.items():
    print(key.capitalize(), ":", value)

with open("study_notes.txt", "a", encoding="utf-8") as file:
    file.write(str(student_notes) + "\n")

print("\nNotes saved successfully to study_notes.txt!")
```
